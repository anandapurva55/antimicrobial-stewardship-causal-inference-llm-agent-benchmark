# Agent Notebook: Antimicrobial Stewardship Causal Study
Runs one trial of a Gemini agent and scores it. Generated code runs in a separate process with a time limit (not `exec`), the model is pinned (no silent fallback) and every artifact (prompt, each attempt, errors, model name, score) is saved under `results/runs/`.

Put `GEMINI_API_KEY` in `.env`.

`CONDITION='original'` leaves the outcome of the mixed-model objective unstated; `'pinned'` names `antibiotic_duration` as the primary outcome for objectives 2 to 4.

```
BACKEND, MODEL = 'gemini', 'gemini-2.5-pro'
CONDITION, TRIAL = 'original', 0   # 'original' | 'pinned'
```

```
from asp_benchmark.agent import run_trial

score = run_trial(BACKEND, MODEL, CONDITION, TRIAL, temperature=0.7, max_attempts=5, exec_timeout=900)
for t in score['tests']:
    print(t['id'], 'PASS' if t['passed'] else 'FAIL', t['msg'])
```

Aggregate across trials with `python evaluation/summarize_runs.py`.
